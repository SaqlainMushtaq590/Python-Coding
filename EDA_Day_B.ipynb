```python
import pandas as pd
```

```python
Data = pd.read_csv("data.csv")
Data
```

```python
Data = Data.drop(["Engine Fuel Type" , "Market Category" , 
                  "Vehicle Style" , "Popularity" , 
                  "Number of Doors" , "Vehicle Size"] , axis = 1)
Data.head()
```

```python
Data.info()
```

```python
Data = Data.rename(columns = {"Engine HP" : "HP" , "Engine Cylinders" : "Cylinders" , 
                              "Transmission Type" : "Transmission" , "Driven_Wheels" : "Drive Mode" ,
                              "highway MPG" : "MPG-H" , "city mpg" : "MPG-C" , "MSRP" : "Price"})
Data.head()
```

```python
Data.shape
```

```python
Duplication = Data[Data.duplicated()]
print("Number of Duplicate Rows: " , Duplication.shape)
```

```python
Data.count()
```

```python
Data = Data.drop_duplicates()
Data.count()
```

```python
print(Data.isnull().sum())
```

```python
Data = Data.dropna()
```

```python
Data.count()
```

```python
print(Data.isnull().sum())
```

```python
Q1 = Data.quantile(0.25)
Q3 = Data.quantile(0.75)
IQR = Q3-Q1
print(IQR)
```

```python
top_makes = Data.Make.value_counts().nlargest(40).reset_index()
top_makes.columns = ["Makes" , "Count"]
figure_2 = px.bar(top_makes, x = "Makes" , y = "Count", 
                  title = "Number of cars by make (Plotly)",
                  labels = {"Count" : "Number of Cars" , "Make" : "Make"})
figure_2.show()
```

```python
top_makes = Data.Make.value_counts().nlargest(20).reset_index()
top_makes.columns = ["Makes" , "Count"]
figure_2 = px.bar(top_makes, x = "Makes" , y = "Count", 
                  title = "Number of cars by make (Plotly)",
                  labels = {"Count" : "Number of Cars" , "Make" : "Make"})
figure_2.show()
```

```python
C = Data.select_dtypes(include = ["number"]).corr()
C
```

```python
figure_3 = px.imshow(C, 
                     text_auto = True, 
                     color_continuous_scale = "BrBG",
                     title = "Correlation Matrix")
figure_3.show()
print(C)
```

```python
plt.figure(figsize = (10,5))
sns.heatmap(C, cmap = "BrBG", annot = True)
plt.show()
```
