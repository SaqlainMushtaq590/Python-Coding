```python
#------------------ ARRAY OPERATIONS----------

# adding arrays
import numpy as np
a = np.array([[2,6,8],
              [2,3,4,]])
b = np.array([[1,2,1],
              [4,3,6]])

c = np.add(a,b)
print(c)


# Subtraction arrays
import numpy as np
a = np.array([[2,6,8],
              [2,3,4,]])
b = np.array([[1,2,1],
              [4,3,6]])

c = np.subtract(a,b)
print(c)


# multipling arrays
import numpy as np
a = np.array([[2,6,8],
              [2,3,4,]])
b = np.array([[1,2,1],
              [4,3,6]])

c = np.multiply(a,b)
print(c)


# dividing  arrays
import numpy as np
a = np.array([[2,6,8],
              [2,3,4,]])
b = np.array([[1,2,1],
              [4,3,6]])

c = np.divide(a,b)
print(c)


# square root of each element
import numpy as np
a = np.array([[2,6,8],
              [2,3,4,]])
b = np.array([[1,2,1],
              [4,3,6]])

print(np.sqrt(a))
print(np.sqrt(b))


# powering each element
import numpy as np
a = np.array([[2,6,8],
              [2,3,4,]])
b = np.array([[1,2,1],
              [4,3,6]])

print(np.power(a,2))
print(" ")
print(np.power(b,3))


#mod of each element by 2
import numpy as np
a = np.array([[2,6,8],
              [2,3,4,]])
b = np.array([[1,2,1],
              [4,3,6]])

print(np.mod(a,2))
print(" ")
print(np.mod(b,3))
import numpy as np
a = np.array([[2,6,8],
              [2,3,4,]])
b = np.array([[1,2,1],
              [4,3,6]])

print(np.clip(a,2,5))
print(" ")
print(np.clip(b,1,3))
```

```python
#-------------- STATISTICAL FUNCTIONS -----------#

import numpy as np

# Array
a = np.array([10, 20, 30, 40, 50])

# Mean
print("Mean:", np.mean(a))

# Median
print("Median:", np.median(a))

# Standard Deviation
print("Standard Deviation:", np.std(a))

# Variance
print("Variance:", np.var(a))

# Minimum value
print("Minimum:", np.min(a))

# Maximum value
print("Maximum:", np.max(a))

# Percentile
print("25th Percentile:", np.percentile(a, 25))
print("50th Percentile:", np.percentile(a, 50))

# Average with weights
weights = [1, 2, 3, 4, 5]
print("Weighted Average:", np.average(a, weights=weights))
```

```python
#---------------------- Aggregate Functions ----------------#

import numpy as np

# Array
a = np.array([1, 2, 3, 4, 5])

# Sum
print("Sum:", np.sum(a))

# Product
print("Product:", np.prod(a))

# Cumulative Sum
print("Cumulative Sum:", np.cumsum(a))

# Cumulative Product
print("Cumulative Product:", np.cumprod(a))

# Index of Maximum value
print("Index of Max Value:", np.argmax(a))

# Index of Minimum value
print("Index of Min Value:", np.argmin(a))

# Boolean array for all() and any()
b = np.array([True, True, False, True])

# Check if all values are True
print("All True?:", np.all(b))

# Check if any value is True
print("Any True?:", np.any(b))
```

```python
#--------------------Linear Algebra  ------------

import numpy as np

# Matrices
a = np.array([[1, 2],
              [3, 4]])

b = np.array([[5, 6],
              [7, 8]])

# Dot Product
print("Dot Product:\n", np.dot(a, b))

# Matrix Multiplication
print("Matrix Multiplication:\n", np.matmul(a, b))

# Inverse of matrix
print("Inverse of a:\n", np.linalg.inv(a))

# Determinant
print("Determinant of a:", np.linalg.det(a))

# Eigenvalues and Eigenvectors
values, vectors = np.linalg.eig(a)
print("Eigenvalues:\n", values)
print("Eigenvectors:\n", vectors)

# Norm
print("Norm of matrix a:", np.linalg.norm(a))

# QR Decomposition
Q, R = np.linalg.qr(a)
print("Q Matrix:\n", Q)
print("R Matrix:\n", R)

# Singular Value Decomposition
U, S, V = np.linalg.svd(a)
print("U Matrix:\n", U)
print("Singular Values:\n", S)
print("V Matrix:\n", V)
```

```python
#------------------------Random Operations  ------------

import numpy as np

# Set seed (same random numbers every run)
np.random.seed(1)

# Random numbers between 0 and 1
print("Random numbers [0,1):", np.random.rand(3))

# Random integers
print("Random Integers:", np.random.randint(1, 10, 5))

# Normal distribution
print("Normal Distribution:", np.random.normal(0, 1, 5))

# Random choice from array
a = np.array([10, 20, 30, 40, 50])
print("Random Choice:", np.random.choice(a, 3))

# Shuffle array
b = np.array([1, 2, 3, 4, 5])
np.random.shuffle(b)
print("Shuffled Array:", b)

# Uniform distribution
print("Uniform Distribution:", np.random.uniform(0, 5, 4))

# Poisson distribution
print("Poisson Distribution:", np.random.poisson(3, 5))
```

```python
#---------------------Shape Operations -----------
import numpy as np

# Original array
a = np.array([[1,2,3],
              [4,5,6]])

print("Original Array:\n", a)

# Reshape
print("\nReshape (3x2):\n", np.reshape(a,(3,2)))

# Flatten
print("\nFlatten:", a.flatten())

# Transpose
print("\nTranspose:\n", np.transpose(a))

# Resize
b = np.array([1,2,3,4])
print("\nResize (2x3):\n", np.resize(b,(2,3)))

# Expand dimensions
c = np.array([1,2,3])
print("\nExpand Dims:\n", np.expand_dims(c, axis=0))

# Squeeze
d = np.array([[[1,2,3]]])
print("\nSqueeze:", np.squeeze(d))

# Ravel
print("\nRavel:", a.ravel())

# Horizontal Stack
x = np.array([1,2,3])
y = np.array([4,5,6])
print("\nHorizontal Stack:", np.hstack((x,y)))

# Vertical Stack
print("\nVertical Stack:\n", np.vstack((x,y)))
```

```python
#-------------------------Indexing & Slicing ------------------

import numpy as np

# Array
a = np.array([10, 20, 10, 30, 40, 0, 50])

print("Original Array:", a)

# np.where()
print("\nIndices where value > 20:", np.where(a > 20))

# np.take()
indices = [0, 2, 4]
print("\nTake elements at indices [0,2,4]:", np.take(a, indices))

# np.unique()
print("\nUnique elements:", np.unique(a))

# np.nonzero()
print("\nIndices of non-zero elements:", np.nonzero(a))

# np.argsort()
print("\nIndices that would sort the array:", np.argsort(a))

# np.split()
b = np.array([1,2,3,4,5,6])
print("\nSplit array:", np.split(b, 3))

# np.flip()
print("\nFlipped array:", np.flip(a))

# np.roll()
print("\nRolled array (shift=2):", np.roll(a, 2))
```
