```python
# adding arrays
import numpy as np
a = np.array([[2,6,8],
              [2,3,4,]])
b = np.array([[1,2,1],
              [4,3,6]])

c = np.add(a,b)
print(c)
```

```python
# Subtraction arrays
import numpy as np
a = np.array([[2,6,8],
              [2,3,4,]])
b = np.array([[1,2,1],
              [4,3,6]])

c = np.subtract(a,b)
print(c)
```

```python
# multipling arrays
import numpy as np
a = np.array([[2,6,8],
              [2,3,4,]])
b = np.array([[1,2,1],
              [4,3,6]])

c = np.multiply(a,b)
print(c)
```

```python
# dividing  arrays
import numpy as np
a = np.array([[2,6,8],
              [2,3,4,]])
b = np.array([[1,2,1],
              [4,3,6]])

c = np.divide(a,b)
print(c)
```

```python
# square root of each element
import numpy as np
a = np.array([[2,6,8],
              [2,3,4,]])
b = np.array([[1,2,1],
              [4,3,6]])

print(np.sqrt(a))
print(np.sqrt(b))
```

```python
# powering each element
import numpy as np
a = np.array([[2,6,8],
              [2,3,4,]])
b = np.array([[1,2,1],
              [4,3,6]])

print(np.power(a,2))
print(" ")
print(np.power(b,3))
```

```python
#mod of each element by 2
import numpy as np
a = np.array([[2,6,8],
              [2,3,4,]])
b = np.array([[1,2,1],
              [4,3,6]])

print(np.mod(a,2))
print(" ")
print(np.mod(b,3))
```

```python
import numpy as np
a = np.array([[2,6,8],
              [2,3,4,]])
b = np.array([[1,2,1],
              [4,3,6]])

print(np.clip(a,2,5))
print(" ")
print(np.clip(b,1,3))
```
